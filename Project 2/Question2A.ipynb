```python
import pandas as pd
```

```python
def extract(filepath):
    df = pd.read_excel("Q2_eCommerce/Online Retail.xlsx", engine='openpyxl')
    print(f"Loaded {len(df)} rows, {len(df.columns)} columns")
    print("Columns:", df.columns.tolist())
    print(df.head(3))
    return df
df = extract("Q2_eCommerce/Online Retail.xlsx")
```

```python
country_to_region = {
    'United Kingdom': 'Europe', 'Germany': 'Europe', 'France': 'Europe',
    'Spain': 'Europe', 'Netherlands': 'Europe', 'Belgium': 'Europe',
    'Portugal': 'Europe', 'Italy': 'Europe', 'Switzerland': 'Europe',
    'Austria': 'Europe', 'Norway': 'Europe', 'Sweden': 'Europe',
    'Finland': 'Europe', 'Denmark': 'Europe', 'Poland': 'Europe',
    'Czech Republic': 'Europe', 'Greece': 'Europe', 'Cyprus': 'Europe',
    'Malta': 'Europe', 'Iceland': 'Europe', 'Lithuania': 'Europe',
    'Latvia': 'Europe', 'Estonia': 'Europe', 'Channel Islands': 'Europe',
    'EIRE': 'Europe', 'RSA': 'Europe', 'European Community': 'Europe',
    
    'USA': 'North America', 'Canada': 'North America',
    
    'Brazil': 'South America',

    'Saudi Arabia': 'Middle East', 'Bahrain': 'Middle East',
    'Lebanon': 'Middle East', 'United Arab Emirates': 'Middle East',

    'Australia': 'Asia Pacific', 'Japan': 'Asia Pacific',
    'Singapore': 'Asia Pacific', 'Hong Kong': 'Asia Pacific',
    'Israel': 'Asia Pacific',

    'Nigeria': 'Africa', 'South Africa': 'Africa',

    'Unspecified': 'Unspecified',
}

def preprocess(df):
    df['Region'] = df['Country'].map(country_to_region)

    unmapped = df[df['Region'].isna()]['Country'].unique()
    if len(unmapped) > 0:
        print("WARNING — unmapped countries:", unmapped)
        df['Region'] = df['Region'].fillna('Other')

    print(f"Total columns: {len(df.columns)}")
    print("Region counts:\n", df['Region'].value_counts())

    region_dfs = {}
    for region in df['Region'].unique():
        region_dfs[region] = df[df['Region'] == region].copy()
        print(f"{region}: {len(region_dfs[region])} rows")

    return df, region_dfs
df, region_dfs = preprocess(df)
```

```python
def persist(region_dfs):
    for region, data in region_dfs.items():
        filename = f"{region.replace(' ', '_')}.csv"
        data.to_csv(filename, index=False)
        print(f"Saved: {filename} — {len(data)} rows")
persist(region_dfs)
```
